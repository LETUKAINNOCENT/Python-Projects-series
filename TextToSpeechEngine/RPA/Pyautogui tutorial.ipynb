```python
import pyautogui
import time
import keyboard
```

```python
try:
    while True:
        x, y = pyautogui.position()
        print(f"x={x}, y={y}", end="\r")
        time.sleep(.25)
except KeyboardInterrupt:
    print("\nDone")
```

```python
try:
    while True:
        x, y = pyautogui.position()
        print(f"x={x}, y={y}", end="\r")
        
        if keyboard.is_pressed('shift+ctrl+alt+k'):
            print("\nGlobal stop triggered!")
            break
            
        time.sleep(.25)
except KeyboardInterrupt:
    print("\nDone via standard interrupt")
finally:
    print("Execution cleanly finished.")
```

```python
pyautogui.moveTo(x,y)
```

```python
time.sleep(4) 
pyautogui.click(x,y,button="right")
```

```python
time.sleep(4) 
pyautogui.click(x,y,button="left")# default
```
