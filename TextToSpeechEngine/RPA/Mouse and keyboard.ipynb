```python
def get_img_position(
    location: str,
    confidence: float = 0.8,
    grayscale: bool = False
) -> Optional[pyautogui.Point]:
    """
    Find an image on the screen and return its center position.

    Args:
        location: Path to the image file.
        confidence: Match confidence (requires OpenCV).
        grayscale: Search in grayscale for faster matching.

    Returns:
        pyautogui.Point if found, otherwise None.
    """

    try:
        return pyautogui.locateCenterOnScreen(
            location,
            confidence=confidence,
            grayscale=grayscale
        )

    except pyautogui.ImageNotFoundException:
        return None

    except Exception as e:
        return None
```

```python
get_img_position("./Assets/light-open-tab.png") 
```

```python
def minimize(dark: bool = True) -> bool:
    theme = "Dark" if dark else "White"

    position = get_img_position(f"./Assets/{theme.lower()}.png")

    if position is None:
        print(f"{theme} image not found")
        return False

    pyautogui.moveTo(position, duration=0)


    pyautogui.moveRel(xOffset=-45, yOffset=0, duration=0)

    pyautogui.click()

    return True
```

```python
minimize()
minimize(False)
```

```python
def close(dark: bool = True) -> bool:
    theme = "Dark" if dark else "White"

    position = get_img_position(f"./Assets/{theme.lower()}.png")

    if position is None:
        print(f"{theme} image not found")
        return False

    pyautogui.moveTo(position, duration=0)


    pyautogui.moveRel(xOffset=45, yOffset=0, duration=0)

    pyautogui.click()

    return True
```

```python
minimize()
close(False)
```
